One Way Anova

```python
import pandas as pd
from scipy.stats import f_oneway

df = pd.read_csv("HW4_Data/Sports.csv")

groups = [df[c].dropna().values for c in df.columns]

print({c: df[c].mean() for c in df.columns})

F, p = f_oneway(*groups)

print(f"F-statistic: {F:.3f}")
print(f"P-value: {p:.4f}")

alpha = 0.05
if p < alpha:
    print("Reject H₀: At least one sport's average income differs.")
else:
    print("Fail to reject H₀: No significant difference in average income among sports.")

#I assume this would also work:

# groups = [
#     df[df["Neighborhood"] == "NAmes"]["SalePrice"],
#     df[df["Neighborhood"] == "CollgCr"]["SalePrice"],
#     df[df["Neighborhood"] == "OldTown"]["SalePrice"]
# ]

# f_stat, p_value = stats.f_oneway(*groups)
# print("F:", f_stat, "p:", p_value)
```

Multiple Regression

```python
#I assume this is right:
#24

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# --- Step 1. Create dataset manually (if not reading from file) ---
data = {
    "Miles": [45, 100, 40, 110, 100],
    "Weight": [15, 25, 20, 30, 35],
    "Speed": [45, 45, 65, 60, 50],
    "Oil": [15, 30, 15, 25, 10]
}

df = pd.DataFrame(data)

# The dependent variable (response) is assumed to be "Time to Overhaul"
# For demonstration, let’s assume you have it as a hypothetical y variable:
# (If not given, you’ll replace this line once your full dataset is available)
df["Overhaul_Time"] = [60, 90, 55, 95, 85]  # Example target variable

# --- Step 2. Split predictors (X) and response (y) ---
X = df[["Miles", "Weight", "Speed", "Oil"]]
y = df["Overhaul_Time"]

# --- Step 3. Fit multiple linear regression model ---
model = LinearRegression()
model.fit(X, y)

# --- Step 4. Regression coefficients ---
print("=== Regression Model ===")
print("Intercept:", model.intercept_)
for col, coef in zip(X.columns, model.coef_):
    print(f"{col}: {coef}")

# --- Step 5. Regression equation ---
print("\nRegression Equation:")
print(f"Overhaul_Time = {model.intercept_:.2f} "
      f"+ ({model.coef_[0]:.2f})*Miles "
      f"+ ({model.coef_[1]:.2f})*Weight "
      f"+ ({model.coef_[2]:.2f})*Speed "
      f"+ ({model.coef_[3]:.2f})*Oil")

# --- Step 6. Model fit ---
y_pred = model.predict(X)
r2 = r2_score(y, y_pred)
print(f"\nModel R² (Goodness of Fit): {r2:.4f}")

# --- Step 7. Plot Actual vs Predicted ---
plt.figure(figsize=(7,5))
plt.scatter(y, y_pred, color="royalblue", edgecolor="black")
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--')
plt.xlabel("Actual Overhaul Time")
plt.ylabel("Predicted Overhaul Time")
plt.title("Actual vs Predicted Engine Overhaul Time")
plt.grid(True)
plt.show()

# --- Step 8. Pairplot for relationships ---
sns.pairplot(df, x_vars=["Miles", "Weight", "Speed", "Oil"], y_vars="Overhaul_Time",
             kind="reg", height=3)
plt.suptitle("Scatter Plots with Regression Lines", y=1.02)
plt.show()
```
